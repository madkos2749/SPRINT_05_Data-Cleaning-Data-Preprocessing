# DATA TYPE HANDLING

### DEFINITION : 
**Data Type Handling** is the process of identifying, managing, converting, and processing different types of data, such as integers, strings, floating-point numbers, characters, and Boolean values, to ensure that data is used correctly in a program or system.

#### EXAMPLE :

```python
age = 24
height = 5.3
name = "Queen"
is_student = False

print(type(age))
print(type(height))
print(type(name))
print(type(is_student))
```

## IDENTIFYING DATA TYPES

### DEFINITION:
**Identifying Data Types** is the process of determining the type of data stored in a variable, such as integer, float, string, or Boolean. It helps the programmer understand what kind of value a variable contains.

#### EXAMPLE:

```python
age = 24
height = 5.3
name = "Queen"
is_student = False

print(type(age))
print(type(height))
print(type(name))
print(type(is_student))
```

## NUMERICAL DATA

### DEFINITION:
**Numerical Data** refers to data that contains numbers used for mathematical calculations. In Python, numerical data is mainly divided into two types: integers (int) and floating-point numbers (float).

#### EXAMPLE:

```python
age = 24
height = 5.3
name = "Queen"
is_student = False

print(age)
print(height)
print(type(age))
print(type(height))
```

## CATEGORICAL DATA

### DEFINITION:
**Categorical Data** refers to data that represents names, labels, or groups rather than numerical values. It is used to classify information into different categories, such as a person's name or student status.

#### EXAMPLE :

```python
age = 24
height = 5.3
name = "Queen"
is_student = True

print(name)
print(is_student)
print(type(name))
print(type(is_student))
```

## BOOLEAN DATA

### DEFINITION:
**Boolean Data** is a data type that represents only two values: True or False. It is used to make decisions and check conditions in a program.

#### EXAMPLE :

```python
age = 24
height = 5.3
name = "Queen"
is_student = True

print(is_student)
print(type(is_student))
```

## DATE/TIME DATA

### DEFINITION:
**Date/Time Data** refers to data that stores dates and time values. It is used to represent a particular day, month, year, and time in a program. In Python, the datetime module is used to handle date and time data.

#### EXAMPLE:

```python
from datetime import datetime

age = 24
height = 5.3
name = "Queen"
is_student = True

current_date = datetime.now()

print(current_date)
print(type(current_date))
```

## STRING DATA

### DEFINITION:
**String Data** refers to data that contains a sequence of characters, such as letters, words, numbers, or symbols. In Python, strings are written inside single or double quotation marks. It is used to store and display text information.

#### EXAMPLE:

```python
age = 24
height = 5.3
name = "Queen"
is_student = True

print(name)
print(type(name))
```

## CONVERTING DATA TYPES

### DEFINITION:
**Converting Data Types** is the process of changing a value from one data type to another, such as converting an integer into a float or a number into a string. In Python, this is called Type Casting.

#### EXAMPLE:

```python
age = 24
height = 5.3
name = "Queen"
is_student = True

age = float(age)
height = int(height)

print(age)
print(height)
print(type(age))
print(type(height))
```

## astype()

### DEFINITION : 
**astype()** is not a standard Python function. The correct function is type(), which is used to identify the data type of a variable. It tells whether a value is an integer, float, string, or Boolean.

#### EXAMPLE:

```python
age = 24
height = 5.3
name = "Queen"
is_student = True

print(type(age))
print(type(height))
print(type(name))
print(type(is_student))
```

## to_numeric()

### DEFINITION:
**to_numeric()** is a function in the Pandas library used to convert values into numerical data types, such as integers or floating-point numbers. It is useful when numerical values are stored as strings.

#### EXAMPLE:

```python
import pandas as pd

age = 24
height = 5.3
name = "Queen"
is_student = True

age = pd.to_numeric(str(age))

print(age)
print(type(age))
```

## to_datetime()

### DEFINITION:
to_datetime() is a function in the Pandas library used to convert date and time values into a proper datetime format. It helps in handling, comparing, and processing dates and times easily.

#### EXAMPLE:

```python
import pandas as pd

age = 24
height = 5.3
name = "Queen"
is_student = True

current_date = pd.to_datetime("2026-11-6")

print(current_date)
print(type(current_date))
```

## HANDLING INVALID CONVERSIONS

### DEFINITION:
**Handling Invalid Conversions** is the process of managing errors that occur when converting data from one type to another. It helps prevent a program from stopping when it encounters incorrect or invalid values. In Python, errors='coerce' can be used with Pandas to handle invalid numerical values by converting them into NaN.

#### EXAMPLE:

```python
import pandas as pd

age = 24
height = 5.3
name = "Queen"
is_student = True

age = pd.to_numeric("twenty", errors='coerce')

print(age)
```

## DETECTING INCORRECT DATA TYPES

### DEFINITION:
**Detecting Incorrect Data Types** is the process of identifying values that have an unexpected or wrong data type in a program. It helps programmers find and correct data errors to ensure proper processing.

#### EXAMPLE:

```python
age = 24
height = 5.3
name = "Queen"
is_student = True

if not isinstance(age, int):
    print("Incorrect data type")
else:
    print("Correct data type")

print(type(age))
```

## String → Numeric

### DEFINITION:
**String to Numeric conversion** is the process of converting a string value (text) containing numbers into a numerical data type, such as integer (int) or float (float). It helps perform mathematical calculations using values stored as strings.

#### EXAMPLE:

```python
age = 24
height = 5.3
name = "Queen"
is_student = True

age = int("24")
height = float("5.3")

print(age)
print(height)
print(type(age))
print(type(height))
```

## String → Date

### DEFINITION:
**String to Date conversion** is the process of converting a date stored as text (string) into a proper date or datetime format. It helps perform date-related operations, such as comparing dates and calculating time differences.

#### EXAMPLE:

```python
from datetime import datetime

age = 24
height = 5.3
name = "Queen"
is_student = True

current_date = datetime.strptime("2026-11-6", "%Y-%m-%d")

print(current_date)
print(type(current_date))
```

## NUMERIC → CATEGORICAL

### DEFINITION:
**Numeric to Categorical conversion** is the process of converting numerical values into groups or categories based on specific conditions. It helps classify numerical data into meaningful groups, such as low, medium, and high.

#### EXAMPLE:

```python
age = 24
height = 5.3
name = "Queen"
is_student = True

if age < 18:
    category = "Minor"
else:
    category = "Adult"

print(category)
print(type(category))
```

## INTEGER → FLOAT

### DEFINITION:
**Integer to Float conversion** is the process of converting a whole number (integer) into a decimal number (float). In Python, the float() function is used to convert an integer into a floating-point number.

#### EXAMPLE:

```python
age = 24
height = 5.3
name = "Queen"
is_student = True

age = float(age)

print(age)
print(type(age))
```
