# OUTLIER TREATMENT

## What is an OUTLIER ?

### DEFINITION:
 An **Outlier** is an unusual value that is much higher or lower than the remaining data values.

* **Why is Outlier Detection important?**
    * Identifies unusual data values.
    * Helps find data entry errors.
    * Improves data accuracy.
    * Helps in better data analysis.

#### EXAMPLE:

```python
marks = [45, 50, 48, 52, 49, 98]

average = sum(marks) / len(marks)

for m in marks:
    if m > average + 20:
        print("Outlier found:", m)
```

**What should we do with a detected outlier?**

* For every detected outlier, we should carefully examine the value before deciding what to do.

* **Example: The detected outlier is 98 marks.**

  * **1. Is it an error?**
Check whether 98 was entered incorrectly. If it is correct, it is not an error.

  * **2. Is it a valid extreme observation?**
Yes. A student scoring 98 when most students score around 50 can be a genuine result.

  * **3. Should it be removed?**
No. If 98 is a genuine mark, removing it would lose useful information.

  * **4. Should it be capped?**
No. Since 98 is already within the valid marks range of 0–100, capping is unnecessary.

  * **5. Should it be transformed?**
No. For this small marks dataset, transformation is not necessary.

## OUTLIER vs ERROR

### DEFINITION:
* **Outlier:** An outlier is a value that is very different from the other values in a dataset. It may be correct or incorrect.

* **Error:** An error is an incorrect value caused by typing mistakes, measurement problems, or data entry mistakes.

### Difference between OUTLIER and ERROR :

* **OUTLIER**
   * Unusaual Data Value.
   * May be valid or Invalid.
   * Can be genuine information.
   * Should be investigated.

* **ERROR**
   * Incorrect Data Value.
   * Unusally needs correction.
   * Caused by a mistake or a problem.
   * Should be corrected if possible.

## IQR METHOD

### DEFINITION:
**IQR Method** identifies unusually high or low values in a dataset using the middle 50% of the data.

###                             Formula:
                                         IQR = Q3 - Q1

#### EXAMPLE:

```python
import numpy as np

marks = [10, 12, 14, 15, 16, 18, 20, 100]

q1 = np.percentile(marks, 25)
q3 = np.percentile(marks, 75)

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

for m in marks:
    if m < lower or m > upper:
        print("Outlier found:", m)
```

**What should we do with the detected outlier?**
##### Detected outlier: 100

* **1. Is it an error?**

Not necessarily. Check whether 100 was entered correctly.

* **2. Is it a valid extreme observation?**

It may be valid if a student genuinely scored 100 marks.

* **3. Should it be removed?**

No, not automatically. Keep it if it is a genuine observation.

* **4. Should it be capped?**

No. If 100 is a valid mark, capping it is unnecessary.

* **5. Should it be transformed?**

Not necessary for this small dataset unless there is a specific analytical reason.

## Z - SCORE METHOD

### DEFINITION:
**Z-Score** tells us how far a value is from the average of a dataset.

**Outlier rule:**

If Z-Score is **greater than +3**, it may be an outlier.

If Z-Score is **less than -3**, it may be an outlier.

If Z-Score is **between -3 and +3**, it is generally not considered an outlier by this rule.

#### EXAMPLE:

```python
import numpy as np

marks = [45, 48, 50, 52, 49, 51, 150]

mean = np.mean(marks)
std = np.std(marks)

for m in marks:
    z = (m - mean) / std
    if abs(z) > 2:
        print("Outlier found:", m)
```

## PERCENTILE METHOD

### DEFINITION:
**Percentile Method** identifies unusually low or high values by checking whether they fall outside a specified percentage range.

* **How does it work?**
   * **Lower Percentile (5th):** Identifies unusually low values.
   * **Upper Percentile (95th):** Identifies unusually high values.
   * Values below the 5th percentile or above the 95th percentile are considered potential outliers.

#### EXAMPLE:

```python
import numpy as np

salaries = [20000, 22000, 25000, 27000,
            30000, 32000, 35000, 40000,
            45000, 50000, 200000]

# Calculate percentiles
lower = np.percentile(salaries, 5)
upper = np.percentile(salaries, 95)

# Detect outliers
outliers = [x for x in salaries
            if x < lower or x > upper]

print("Lower Limit:", lower)
print("Upper Limit:", upper)
print("Outliers:", outliers)
```

## WINSORIZATION

### DEFINITION:
**Winsorization** limits unusually small and large values to reduce their influence on statistical analysis and machine learning models.

#### EXAMPLE:

```python
import numpy as np
from scipy.stats.mstats import winsorize

# Product delivery times in minutes
times = [20, 22, 25, 28, 30, 32, 35, 40, 150]

# Apply Winsorization
result = winsorize(times, limits=[0.1, 0.1])

print("Original Data:", times)
print("Winsorized Data:", result)
```

## CAPPING

### DEFINITION:
**Capping** is an outlier-handling technique in which extreme values are replaced with predefined minimum and maximum boundary values.

#### EXAMPLE:

```python
import numpy as np

# Monthly sales in rupees
sales = [500, 800, 1000, 1200, 1500, 9000]

# Define boundaries
lower = 500
upper = 2000

# Apply capping
capped_sales = np.clip(sales, lower, upper)

print("Original:", sales)
print("After Capping:", capped_sales)
```

## FLOORING

### DEFINITION:
**Flooring** prevents unusually small values from going below an acceptable minimum limit.

Values below the lower limit are replaced with the lower limit.

Values above the lower limit remain unchanged.

#### EXAMPLE:

```python
import numpy as np

# Monthly expenses
expenses = [500, 800, 1200, 1500, 2000, 3000]

# Define minimum limit
floor = 1000

# Apply flooring
result = np.maximum(expenses, floor)

print("Original:", expenses)
print("After Flooring:", result)
```

## TRANSFORMATION

### DEFINITION:
**Transformation** changes the values mathematically without deleting the original observations.

* It is especially useful when numerical data is highly skewed.

* Common transformation techniques include:

   * Log Transformation

   * Square Root Transformation

   * Cube Root Transformation

   * Box-Cox Transformation

#### EXAMPLE:

```python
import numpy as np

# Monthly product sales
sales = [100, 200, 300, 400, 500, 10000]

# Apply log transformation
result = np.log1p(sales)

print("Original:", sales)
print("Transformed:", np.round(result, 2))
```

## REMOVING OUTLIERS

### DEFINITION:
 In **Removing Outliers**, we identify extreme values, investigate them, and remove only those that should not be part of the dataset.

#### EXAMPLE:

```python
import pandas as pd

# Student marks
marks = pd.Series([45, 50, 55, 60, 65, 70, 75, 500])

# Calculate IQR boundaries
Q1 = marks.quantile(0.25)
Q3 = marks.quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Identify potential outliers
outliers = marks[(marks < lower) | (marks > upper)]

# Remove confirmed invalid values
cleaned = marks[~marks.isin(outliers)]

print("Outliers:", outliers.tolist())
print("Cleaned Data:", cleaned.tolist())
```

## RETAINING OUTLIERS

### DEFINITION:
**Retaining Outliers** is a technique in which unusual or extreme observations are kept in the dataset instead of removing, capping, or transforming them.

#### EXAMPLE:

```python
import pandas as pd

# Student marks
marks = pd.Series([45, 50, 55, 60, 65, 70, 75, 100])

# Detect potential outliers using IQR
Q1 = marks.quantile(0.25)
Q3 = marks.quantile(0.75)
IQR = Q3 - Q1

upper = Q3 + 1.5 * IQR

outliers = marks[marks > upper]

# Retain all observations
retained_data = marks.copy()

print("Detected Outliers:", outliers.tolist())
print("Retained Data:", retained_data.tolist())
```
