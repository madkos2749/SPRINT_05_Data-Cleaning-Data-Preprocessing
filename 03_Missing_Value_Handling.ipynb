# MISSING VALUE HANDLING

## What are MISSING VALUES ?

### DEFINITION:
**Missing Values** are data entries that are not available or have not been recorded in a dataset. They are represented by empty cells, None, NaN, or blank spaces. Missing values can affect data analysis and produce incorrect results if they are not handled properly.

#### EXAMPLE:

```python
import pandas as pd

data = {
    "Name": ["Arpitha", "Ankitha", "Anjitha", "Aparna"],
    "Age": [20, None, 22, 23]
}

df = pd.DataFrame(data)

print(df)
```

## Why MISSING VALUES occur ?

### DEFINITION:
**Missing Values** occur when some data entries are not available, not recorded, or lost during data collection and processing. They create empty spaces in a dataset and may affect the accuracy of data analysis.

* **Reasons Why Missing Values Occur :**

  * **1. Data Entry Errors:** Users may forget to enter some information.
  * **2. Human Errors:** Incorrect data entry or accidental deletion can cause missing values.
  * **3. Data Collection Problems:** Some information may not be collected properly.
  * **4. User Choice:** People may refuse to provide personal information.
  * **5. Technical Issues:** System failures or software errors may cause data loss.
  * **6. Data Transfer Errors:** Data may be lost while transferring between systems.

#### EXAMPLE:

```python
import pandas as pd

data = {
    "Name": ["Arpitha", "Ankitha", "Anjitha", "Aparna"],
    "Age": [20, None, 22, 23]
}

df = pd.DataFrame(data)

print(df)
```

## MCAR (MISSING COMPLETELY AT RANDOM)

### DEFINITION:
**MCAR** stands for Missing Completely At Random. It is a type of missing data where the missing values occur randomly and have no relationship with any other data or variables in the dataset. The missing values happen due to unexpected reasons, such as technical errors or accidental data loss.

#### EXAMPLE:
If a person's age is missing because of a temporary computer error, it is an example of MCAR.

```python
import pandas as pd

data = {
    "Name": ["Arpitha", "Ankitha", "Anjitha", "Aparna"],
    "Age": [20, None, 22, 23]
}

df = pd.DataFrame(data)

print(df)
```

## MAR (MISSING AT RANDOM)

### DEFINITION:
**MAR** stands for Missing At Random. It is a type of missing data where the missing values depend on other available information or variables in the dataset, but not on the missing value itself.

#### EXAMPLE:
If younger people are more likely to leave their age information blank, and their age group is known from another available variable, it can be an example of MAR.

```python
import pandas as pd

data = {
    "Name": ["Arpitha", "Ankitha", "Anjitha", "Aparna"],
    "Age": [20, None, 22, 23]
}

df = pd.DataFrame(data)

print(df)
```

## MNAR (MISSING NOT AT RANDOM)

### DEFINITION:
**MNAR** stands for Missing Not At Random. It is a type of missing data where the probability of a value being missing depends on the value itself. In this case, the missing data is related to the actual information that is not provided.

#### EXAMPLE:
If people with very high incomes intentionally avoid sharing their income details, it is an example of MNAR.

```python
import pandas as pd

data = {
    "Name": ["Arpitha", "Ankitha", "Anjitha", "Aparna"],
    "Age": [20, None, 22, 23]
}

df = pd.DataFrame(data)

print(df)
```

## DETECTING MISSING VALUES

### DEFINITION:
**Detecting missing values** means identifying empty or unavailable data in a dataset.

* In Pandas, we use the following functions to detect missing values:

  * **isnull()** – Returns True if a value is missing.
  * **notnull()** – Returns True if a value is not missing.
  * **isnull().sum()** – Counts the missing values in each column.

#### EXAMPLE:

```python
# Detect missing values
print(df.isnull())

# Count missing values
print(df.isnull().sum())
```

## MISSING VALUE PERCENTAGE

### DEFINITION:
**Missing Value Percentage** means calculating how much data is missing in each column of a dataset in percentage form.

* It helps us understand which columns have more missing data and need attention.

**Formula :**

              
               
                                    Missing Values
              Missing Percentage = ---------------- * 100
                                     Total Values

#### EXAMPLE:

```python
# Calculate missing value percentage
percentage = (df.isnull().sum() / len(df)) * 100

print("\nMissing Value Percentage:")
print(percentage)
```

## DROPPING ROWS

### DEFINITION:
**Dropping Rows** means removing entire rows from a dataset that contain missing values.

* In Pandas, we use the dropna() function to remove rows containing missing values.

#### EXAMPLE:

```python
# Remove rows with missing values
df.dropna(inplace=True)

print("\nAfter Dropping:")
print(df)
```

## DROPPING COLUMNS

### DEFINITION:
**Dropping Columns** means removing an entire column from a dataset, usually when it contains too many missing values or is not useful for analysis.

* In Pandas, we use the drop() function with columns to remove a column.

#### EXAMPLE:

```python
print("Before Dropping:")
print(df)

# Remove Age column
df.drop(columns=["Age"], inplace=True)

print("\nAfter Dropping:")
print(df)
```

## MEAN IMPUTATION

### DEFINITION:
**Mean Imputation** is a technique used to fill missing numerical values with the average (mean) of the available values in that column.

* It helps us handle missing data without deleting entire rows.

**Formula :**

                                         Sum of Values
                                  Mean = ----------------
                                        Number of Values

#### EXAMPLE:

```python
df = pd.DataFrame(data)
print("Before Imputation:")
print(df)

# Replace missing Age with mean
df["Age"] = df["Age"].fillna(df["Age"].mean())

print("\nAfter Mean Imputation:")
print(df)
```

## MEDIAN IMPUTATION

### DEFINITION:
**Median Imputation** is a technique used to fill missing numerical values with the median (middle value) of the available values in that column.

* It is useful when a dataset contains outliers because the median is less affected by extreme values than the mean.

* **Formula :**
    * First, arrange the values in ascending order and find the middle value.

For example:

20, 22, 23

Median = 22

#### EXAMPLE:

```python
df = pd.DataFrame(data)

print("Before Imputation:")
print(df)

# Replace missing Age with median
df["Age"] = df["Age"].fillna(df["Age"].median())

print("\nAfter Median Imputation:")
print(df)
```

## MODE IMPUTATION

### DEFINITION:
**Mode Imputation** is a technique used to fill missing values with the most frequently occurring value in a column.

* It is mainly used for categorical data, but it can also be used for numerical data.

#### EXAMPLE:

```python
df = pd.DataFrame(data)

print("Before Imputation:")
print(df)

# Replace missing Age with mode
df["Age"] = df["Age"].fillna(df["Age"].mode()[0])

print("\nAfter Mode Imputation:")
print(df)
```

## CONSTANT VALUE IMPUTATION

### DEFINITION:
**Constant Value Imputation** is a technique used to fill missing values with a fixed value chosen by the user instead of using mean, median, or mode.

* For example, we can replace missing Age values with 0 or -1.

#### EXAMPLE:

```python
df = pd.DataFrame(data)

print("Before Imputation:")
print(df)

# Replace missing Age with constant value 18
df["Age"] = df["Age"].fillna(18)

print("\nAfter Constant Imputation:")
print(df)
```

## FORWARD FILL (ffill)

### DEFINITION:
Forward Fill is a technique used to replace missing values with the previous available value in the same column.

* In Pandas, we use the ffill() function to fill missing values from the previous row.

#### EXAMPLE:

```python
df = pd.DataFrame(data)

print("Before Forward Fill:")
print(df)

# Fill missing value with previous value
df["Age"] = df["Age"].ffill()

print("\nAfter Forward Fill:")
print(df)
```

## BACKWARD FILL [bfill]

### DEFINITION:
Backward Fill is a technique used to replace missing values with the next available value in the same column.

* In Pandas, we use the bfill() function to fill missing values using the value from the next row.

#### EXAMPLE:

```python
df = pd.DataFrame(data)

print("Before Backward Fill:")
print(df)

# Fill missing value with next value
df["Age"] = df["Age"].bfill()

print("\nAfter Backward Fill:")
print(df)
```

## INTERPOLATION

### DEFINITION:
Interpolation is a technique used to fill missing numerical values by estimating them from the available values before and after the missing value.

* In Pandas, we use the interpolate() function to estimate missing values.

#### EXAMPLE:

```python
df = pd.DataFrame(data)

print("Before Interpolation:")
print(df)

# Fill missing value using interpolation
df["Age"] = df["Age"].interpolate()

print("\nAfter Interpolation:")
print(df)
```

## GROUP-BASED IMPUTATION

### DEFINITION:
**Group-Based Imputation** is a technique used to fill missing values using the mean, median, or mode of a particular group instead of using the entire dataset.

* It is useful when different groups have different data patterns.

#### EXAMPLE:
Suppose students belong to different groups. We fill a missing Age using the average Age of students in the same group.

```python
import pandas as pd

data = {
    "Name": ["Arpitha", "Ankitha", "Anjitha", "Aparna"],
    "Age": [20, None, 22, 23],
    "Group": ["A", "A", "B", "B"]
}

df = pd.DataFrame(data)

print("Before Imputation:")
print(df)

# Fill missing Age using group mean
df["Age"] = df.groupby("Group")["Age"].transform(
    lambda x: x.fillna(x.mean())
)

print("\nAfter Group-Based Imputation:")
print(df)
```

## KNN IMPUTATION

### DEFINITION:
**KNN Imputation** is a technique used to fill missing values by looking at the values of the nearest similar data points (neighbors).

* It finds similar records and uses their values to estimate the missing value.

#### EXAMPLE:
If Ankitha's Age is missing, KNN looks at the nearest similar records and uses their Age values to estimate her missing Age.

```python
import pandas as pd
from sklearn.impute import KNNImputer

data = {
    "Name": ["Arpitha", "Ankitha", "Anjitha", "Aparna"],
    "Age": [20, None, 22, 23]
}

df = pd.DataFrame(data)

print("Before Imputation:")
print(df)

# Apply KNN Imputation
imputer = KNNImputer(n_neighbors=2)
df[["Age"]] = imputer.fit_transform(df[["Age"]])

print("\nAfter KNN Imputation:")
print(df)
```

## ITERATIVE IMPUTATION

### DEFINITION:
Iterative Imputation is a technique used to fill missing values by repeatedly predicting them using other available columns in the dataset.

#### EXAMPLE:
If Ankitha's Age is missing, Iterative Imputation uses other available information to predict her Age and gradually improves the estimate.

```python
import pandas as pd
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

data = {
    "Name": ["Arpitha", "Ankitha", "Anjitha", "Aparna"],
    "Age": [20, None, 22, 23]
}

df = pd.DataFrame(data)

print("Before Imputation:")
print(df)

# Apply Iterative Imputation
imputer = IterativeImputer(random_state=10)
df[["Age"]] = imputer.fit_transform(df[["Age"]])

print("\nAfter Iterative Imputation:")
print(df)
```
