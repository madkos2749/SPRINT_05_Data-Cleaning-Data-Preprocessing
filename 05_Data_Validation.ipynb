# DATA VALIDATION

## What is Data Validation ?

### DEFINITION:
**Data Validation** is the process of checking whether the data entered by a user is correct, valid, complete, and follows the required rules before accepting or storing it.

* **Data Validation is Important for :**
    * Prevents incorrect data entry.
    * Ensures data accuracy.
    * Avoids missing or invalid information.
    * Improves data quality and reliability.

#### EXAMPLE:

```python
email = input("Enter your email: ")

if len(email) >= 10 and "@" in email and "." in email:
    print("Valid email")
else:
    print("Invalid email")
```

## RANGE VALIDATION

### DEFINITION:
**Range Validation** ensures that the entered value is within the allowed range.

* **Why is Range Validation important?**

   * Prevents values that are too high or too low.
   * Avoids incorrect data entry.
   * Maintains data accuracy.
   * Ensures values follow given limits.

#### EXAMPLE:

```python
marks = int(input("Enter your marks: "))

if 0 <= marks <= 100:
    print("Accepted: Valid marks")
else:
    print("Rejected: Invalid marks")
```

## TYPE VALIDATION

### DEFINITION:
**Type Validation** ensures that the user enters the correct type of data.

* **Why is Type Validation important?**
   * Prevents incorrect data types.
   * Avoids program errors.
   * Maintains data accuracy.
   * Ensures correct input for processing.

#### EXAMPLE:

```python
temp = input("Enter temperature: ")

if temp.isdigit():
    print("Valid: Temperature accepted")
else:
    print("Invalid: Enter numbers only")
```

## BUSINESS RULE VALIDATION

### DEFINITION:
**Business Rule Validation** ensures that data follows the specific rules of a business.

* **Why is Business Rule Validation important?**
    * Ensures business rules are followed.
    * Prevents invalid transactions.
    * Reduces errors in business operations.
    * Maintains data accuracy and consistency.

#### EXAMPLE:

## REFERENTIAL VALIDATION

### DEFINITION:
**Referential Validation** ensures that the entered data refers to an existing record in another table.

* **Why is Referential Validation important?**
     * Maintains relationships between tables.
     * Prevents invalid references.
     * Avoids missing or incorrect records.
     * Ensures data consistency.

#### EXAMPLE:

```python
student_ids = [101, 102, 103, 104]

sid = int(input("Enter Student ID: "))

if sid in student_ids:
    print("Valid: Student ID exists")
else:
    print("Invalid: Student ID not found")
```

## DATE VALIDATION

### DEFINITION:
**Date Validation** ensures that the entered date is valid and acceptable.

* **Why is Date Validation important?**
     * Prevents incorrect date entry.
     * Ensures dates follow the correct format.
     * Avoids invalid dates.
     * Maintains data accuracy.

#### EXAMPLE:

```python
from datetime import datetime

date = input("Enter date (DD-MM-YYYY): ")

try:
    datetime.strptime(date, "%d-%m-%Y")
    print("Valid Date", date)
except ValueError:
    print("Invalid Date")
```

## CONSTRAINT VALIDATION

### DEFINITION:
**Constraint Validation** ensures that the entered data follows the given rules and limitations.

* **Why is Constraint Validation important?**
     * Prevents invalid data entry.
     * Ensures predefined conditions are followed.
     * Maintains data accuracy.
     * Avoids incorrect information.

#### EXAMPLE:

```python
password = input("Enter your password: ")

if len(password) >= 8:
    print("Valid: Password accepted", password)
else:
    print("Invalid: Password must contain 8 characters")
```
