# **DATA PREPROCESSING FUNDAMENTALS**

## What is Data Preprocessing ?
**Data Preprocessing** means preparing raw data so that it can be easily understood and used by a computer.

#### EXAMPLE:
Suppose a bank collects customer names, but some names contain extra spaces and inconsistent capitalization.

```python
# Raw customer names
customers = ["  ram Charan", "Piyush  ", "  amritha  ", "Fathima   "]

# Preprocess customer names
clean_names = [name.strip().title() for name in customers]

# Display results
print("Raw Data:", customers)
print("Processed Data:", clean_names)
```

## Why is Data Preprocessing required ?

**Data preprocessing** is needed to make raw data clean, understandable, and ready to use.

* **Why do we need Data Preprocessing?**
  
   * **Remove Errors:** Corrects incorrect or unwanted data.
   * **Handle Missing Values:** Manages incomplete information.
   * **Remove Extra Spaces:** Makes text clean and readable.
   * **Maintain Consistency:** Converts data into a standard format.
   * **Improve Accuracy:** Helps produce more reliable results.
   * **Better Performance:** Makes data suitable for machine learning and AI systems.

#### EXAMPLE:

```python
# Raw customer deposit data
deposits = [5000, 3000, -1000, 5000, 2000]

# Process data by removing invalid and duplicate values
valid_deposits = list(set(x for x in deposits if x > 0))

# Calculate total deposits
total = sum(valid_deposits)

print("Valid Deposits:", sorted(valid_deposits))
print("Total Deposits:", total)
```

## DATA QUALITY

### DEFINITION:
**Data Quality** refers to how accurate, complete, consistent, and reliable the data is before using it for analysis or machine learning.

* ### Importance of Data Quality :
    * **Accuracy:** Ensures data contains correct values.
    * **Completeness:** Helps identify and handle missing values.
    * **Consistency:** Ensures data follows the same format.
    * **Reliability:** Makes data trustworthy for analysis.
    * **Better Results:** Improves the performance of machine learning models.

#### EXAMPLE:

```python
import pandas as pd

data = {
    'Name': ['Ram Charan', 'Piyush', 'Amritha', 'Fathima', 'Gunesh'],
    'Age': [20, None, 22, None, 21],
    'Marks': [85, 90, None, 88, 77]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

print("\nData After Cleaning:")
print(df)
```

  * **Improve Reliability:** Ensures the model learns from proper training data.
  * **Better Evaluation:** Provides realistic model performance.
  * **Prevent Overfitting:** Reduces the risk of learning information that should not be available.
  * **Trustworthy Results:** Helps build reliable machine learning models.

## PREPROCESSING PIPELINE

### DEFINITION:
**Preprocessing Pipeline** refers to a sequence of steps used to clean, transform, and prepare raw data before providing it to a machine learning model. It ensures that data is accurate, consistent, complete, and ready for analysis.

* **Explanation of Each Step :**

    * **Raw Data Collection:** Collect raw data from files, databases, websites, or other sources.
    * **Data Cleaning:** Identify and handle missing values, duplicate records, and incorrect information.
    * **Data Transformation:** Convert data into a suitable format using encoding, normalization, or scaling.
    * **Data Splitting:** Divide the dataset into training, validation, and testing data.
    * **Data Validation:** Check whether the processed data follows the required rules and conditions.
    * **Preprocessed Data:** The final cleaned and transformed data is ready for machine learning.
