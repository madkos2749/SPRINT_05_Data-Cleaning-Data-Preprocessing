# DUPLICATE DATA HANDLING

**Duplicate Data Handling** is the process of identifying and removing repeated records from a dataset.

* Duplicate data means the same information appears more than once.
* It can affect data accuracy and produce incorrect analysis results.

* **In Pandas, we use the following functions :**

   * duplicated() – Identifies duplicate rows.
   * drop_duplicates() – Removes duplicate rows.
   * keep="first" – Keeps the first occurrence and removes the remaining duplicates.

## What are Duplicate Records ?

### DEFINITION:
Duplicate Records are rows in a dataset that contain the same information as another row.
* Duplicate records can reduce data quality and affect the accuracy of data analysis.

#### EXAMPLE:

```python
import pandas as pd

data = {
    "Product": ["Laptop", "Mobile", "Tablet", "Laptop"],
    "Price": [50000, 20000, 15000, 50000]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

# Find duplicate records
print("\nDuplicate Records:")
print(df[df.duplicated()])
```

## EXACT RECORDS

### DEFINITION:
**Exact Duplicates** are records in a dataset where all column values are exactly the same as another record.

#### EXAMPLE:

```python
print("Original Data:")
print(df)

# Find exact duplicates
print("\nExact Duplicates:")
print(df[df.duplicated(keep=False)])
```

## PARTIAL DUPLICATES

### DEFINITION:
**Partial Duplicates** are records in a dataset where only some column values are repeated, while other column values are different.

#### EXAMPLE:

```python
print("Original Data:")
print(df)

# Find partial duplicates based on Product
print("\nPartial Duplicates:")
print(df[df.duplicated(subset=["Product"], keep=False)])
```

## IDENTIFYING DUPLICATES

### DEFINITION:
**Identifying Duplicates** means finding repeated records in a dataset to identify duplicate entries.

* In Pandas, we use the duplicated() function to check whether rows are repeated.

#### EXAMPLE:

```python
print("Original Data:")
print(df)

# Identify duplicate products
print("\nDuplicate Records:")
print(df[df.duplicated(subset=["Product"])])
```

## duplicated()

### DEFINITION:
**duplicated()** is a Pandas function used to identify duplicate rows or repeated records in a dataset.

* It returns True for duplicate rows and False for non-duplicate rows.

#### EXAMPLE:

```python
print("Original Data:")
print(df)

# Check duplicate products
print("\nDuplicate Status:")
print(df.duplicated(subset=["Product"]))
```

## drop_duplicates()

### DEFINITION:
**drop_duplicates()** is a Pandas function used to remove repeated records from a dataset.

* It removes duplicate rows and keeps only one occurrence of each repeated record.

#### EXAMPLE:

```python
print("Before Removing Duplicates:")
print(df)

# Remove duplicate rows
df.drop_duplicates(inplace=True)

print("\nAfter Removing Duplicates:")
print(df)
```

## DUPLICATE DETECTION BASED ON SELECTED COLUMNS

### DEFINITION:
Duplicate Detection Based on Selected Columns means identifying duplicate rows in a Pandas DataFrame by checking only specific columns instead of checking all columns.

* In Pandas, we use the duplicated() method with the subset parameter to detect duplicates based on selected columns.

#### EXAMPLE:

```python
print("Original Data:")
print(df)

# Remove duplicates based on Product column
df = df.drop_duplicates(subset=["Product"])

print("\nData After Removing Duplicates:")
print(df)
```

## HANDLING DUPLICATE RECORDS

### DEFINITION:
**Handling Duplicate Records** means identifying and removing duplicate rows from a Pandas DataFrame to maintain clean, accurate, and consistent data.

* In Pandas, we use the drop_duplicates() method to remove duplicate records from a DataFrame.

#### EXAMPLE:

```python
print("Original Data:")
print(df)

# Remove duplicate records
df = df.drop_duplicates()

print("\nData After Removing Duplicates:")
print(df)
```

## BUSINESS RULES FOR DUPLICATE REMOVAL

### DEFINITION:
Business Rules for Duplicate Removal means defining specific conditions or guidelines to decide which duplicate records should be removed and which records should be retained based on business requirements.

* For example:

  * If the same product appears multiple times with the same price, remove the duplicate record.
  * If the same product appears with different prices, retain the record with the highest price or the latest updated price, depending on business requirements.
  * Always retain the most relevant or accurate record.

#### EXAMPLE:

```python
print("Original Data:")
print(df)

# Business Rule:
# If a product appears multiple times,
# keep the record with the highest price.

df = df.sort_values("Price", ascending=False)

df = df.drop_duplicates(subset=["Product"], keep="first")

print("\nData After Applying Business Rules:")
print(df)
```

## Explain why blindly deleting duplicates can sometimes be dangerous.

* **Blindly deleting duplicates** means removing repeated records without checking whether those records contain important or different business information.
* In real-world applications, not all duplicate-looking records are actual duplicates. Sometimes, two records may have the same product name but different prices, dates, quantities, or customer details.
* If we delete such records without understanding the business requirements, we may lose important information and make incorrect business decisions.
